```python
START_DATE = '2026/04/30'
END_DATE = '2026/05/31'
```

```python
import pathlib
import json

with open('config.json') as f:
    config = json.load(f)
download_root = pathlib.Path(config['download_cache_path'])

cooldown_time = 2
interations = ['1st', '2nd']
```

# TWSE  (上市)

## Technicals（技術分析）

### Individual - price and volume （個股 - 價量）

```python
import core.scraper.twse.technicals
import core.scraper.utils

import core.checker.twse.technicals
import core.checker.utils

save_path = download_root.joinpath('twse','technicals','stocks','url_0')

while True:
    core.scraper.utils.download_raw_material(
        core.scraper.twse.technicals.stocks.price_volume.url_0,
        START_DATE,
        END_DATE,
        save_path,
        interations,
        cooldown_time
    )

    stop = core.checker.utils.check_raw_material(
        core.checker.twse.technicals.stocks.price_volume.url_0,
        save_path,
        interations[0],
        interations[1]
    )

    if stop:
        break

    core.checker.utils.remove_inconsistent_raw_material(
        core.checker.twse.technicals.stocks.price_volume.url_0,
        save_path,
        interations[0],
        interations[1]
    )
```
