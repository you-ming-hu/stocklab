```python
import pathlib
import json

root = pathlib.Path.cwd().parent
with open(root.joinpath('config.json')) as f:
    config = json.load(f)

init_root = root.joinpath(config['init_root_path'])

end_date = '2026/04/30'

cooldown_time = 2
interations = ['1st', '2nd']
```

# Technicals information (技術面-價量)

## TWSE (上市)

### Individual stocks (個股)

```python
import scraper.technicals.twse
import scraper.utils

import checker.technicals.twse
import checker.utils

save_path = init_root.joinpath('technicals', 'TWSE','stocks')

while True:
    scraper.utils.download_raw_material(
        scraper.technicals.twse.stocks,
        '2004/02/11',
        end_date,
        save_path,
        interations,
        cooldown_time
    )

    stop = checker.utils.check_raw_material(
        checker.technicals.twse.stocks,
        save_path,
        interations[0],
        interations[1]
    )

    if stop:
        break

    checker.utils.remove_inconsistent_raw_material(
        checker.technicals.twse.stocks,
        save_path,
        interations[0],
        interations[1]
    )
```
