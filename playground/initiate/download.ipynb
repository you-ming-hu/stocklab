```python
import pathlib
import json

root = pathlib.Path.cwd().parent
with open(root.joinpath('config.json')) as f:
    config = json.load(f)

init_root = root.joinpath(config['init_root_path'])

end_date = '2026/04/30'

cooldown_time = 2
interations = ['1st', '2nd']
```

# TWSE  (上市)

## Technicals（技術分析）

### Individual - price and volume （個股 - 價量）

```python
import scraper.twse.technicals
import scraper.utils

import checker.twse.technicals
import checker.utils

save_path = init_root.joinpath('twse','technicals','stocks')

while True:
    scraper.utils.download_raw_material(
        scraper.twse.technicals.stocks.price_volume.version_0,
        '2004/02/11',
        end_date,
        save_path,
        interations,
        cooldown_time
    )

    stop = checker.utils.check_raw_material(
        checker.twse.technicals.stocks.price_volume.version_0,
        save_path,
        interations[0],
        interations[1]
    )

    if stop:
        break

    checker.utils.remove_inconsistent_raw_material(
        checker.twse.technicals.stocks.price_volume.version_0,
        save_path,
        interations[0],
        interations[1]
    )
```
