```python
import database

import pathlib
import json

root = pathlib.Path.cwd().parent
with open(root.joinpath('config.json')) as f:
    config = json.load(f)

database_path = root.joinpath(config['database_path'])
saver = database.commander.Saver(database_path)

init_root = root.joinpath(config['init_root_path'])
```

```python
source_path = init_root.joinpath('technicals','TWSE','stocks','1st')

item = database.sources.technicals.twse.stocks
item.add_path(source_path)

with saver.register(item):
    # saver.init_table()
    cmds = saver.update_table()
    saver.update_database()
```
