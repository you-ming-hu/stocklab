```python
import json

import checker.twse
import checker.utils

import scraper.twse
import scraper.utils

save_path = init_root.joinpath('TWSE','stocks')

stop = False
skip = set()
while not stop:
    
    checker.utils.check_intersection(
        checker.twse.stocks,
        save_path,
        '1st',
        '3rd',
        skip
    )

    checker.utils.check_intersection(
        checker.twse.stocks,
        save_path,
        '2nd',
        '3rd',
        skip
    )

    with open(save_path.joinpath('diff.json')) as f:
        diff = json.load(f)
    
    A = set(diff['1st,2nd'])
    B = set(diff['1st,3rd'])
    C = set(diff['2nd,3rd'])

    invB = A-B
    invC = A-C
    unqualified = A-(invB|invC)
    skip = A - unqualified

    print(f'unqualified count: {len(unqualified)}')

    for q in sorted(unqualified):
        save_path.joinpath('3rd',q).with_suffix(scraper.twse.stocks.suffix).unlink()

    scraper.utils.download_group_material(
        scraper.twse.stocks,
        save_path,
        '3rd',
        sorted(unqualified),
        cooldown_time
    )

    if len(unqualified) == 0:
        stop = True
```
