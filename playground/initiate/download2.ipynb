```python
import pathlib
import json

root = pathlib.Path.cwd().parent
with open(root.joinpath('config.json')) as f:
    config = json.load(f)

init_root = root.joinpath(config['init_root_path'])

cooldown_time = 2
interations = ['1st']
```

# Overview table（總表）Industry（產業別）

```python
MIN_SLEEP_TIME = 0.1
MAX_SLEEP_TIME = 1
```

```python
# TWSE

import scraper.twse.company
import scraper.utils

save_path = init_root.joinpath('twse','company','overview')
scraper.twse.company.overview.version_0.MIN_SLEEP_TIME = 0
scraper.twse.company.overview.version_0.MAX_SLEEP_TIME = 0.1

scraper.utils.download_raw_material(
    scraper.twse.company.overview.version_0,
    None,
    None,
    save_path,
    interations,
    cooldown_time
)

scraper.twse.company.industry.version_0.find_latest_company_table(
    save_path.joinpath(interations[0]),
    scraper.twse.company.overview.version_0.suffix
)

save_path = init_root.joinpath('twse','company','industry')
scraper.twse.company.industry.version_0.MIN_SLEEP_TIME = MIN_SLEEP_TIME
scraper.twse.company.industry.version_0.MAX_SLEEP_TIME = MAX_SLEEP_TIME

scraper.utils.download_raw_material(
    scraper.twse.company.industry.version_0,
    None,
    None,
    save_path,
    interations[:1],
    cooldown_time
)
```
